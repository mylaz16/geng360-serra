# Part 2: Arduino Data Collection

```python
# import the dataset, view the first few rows
myLab = pd.read_csv("Lab2.0data.csv") # Make sure to update this file name to match your data! 
myLab.head() #This will display the first few rows of the DataFrame
```

```python
myLab['Sensor 1 (V)']= myLab['Sensor 1 (raw)'] * (5.06/1023.0) # Convert raw sensor data to volts using the provided conversion factor

plt.figure()  # Create a new figure
sns.lineplot(x="Time (ms)", y="Sensor 1 (V)", data=myLab)
plt.title("Sensor 1 vs Time")
plt.xlabel("Time (ms)")
plt.ylabel("Sensor 1 (V)")

mplcursors.cursor(hover=False)  # Enable interactive cursor for data point selection
# Click anywhere on your data to add a cursor
# Right-click to remove a cursor

plt.show()
```
