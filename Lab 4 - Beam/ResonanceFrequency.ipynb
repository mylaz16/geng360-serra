```python
myLab = pd.read_csv("harderdata.csv") # Load dataset 
myLab.head() #This will display the first few rows of the DataFrame

#Covert raw Arduino data to volts 
myLab['Sensor 1 (V)']= myLab['Sensor 1 (raw)'] * (5.06/1023.0) # Convert raw sensor data to volts using the provided conversion factor

#Subtract the zero g bias so signal is centered around 0V
myLab['Sensor 1 (V)'] = myLab['Sensor 1 (V)'] - 1.65 # Subtract the zero-g bias (2.5V) from the sensor data to center it around 0V

#Convert volts to g's using the sensitivity of the sensor
myLab['Sensor 1 (g)'] = myLab['Sensor 1 (V)'] / 0.340 # Convert volts to g's using the sensitivity of the sensor (0.330 V/g) 

plt.figure()  # Create a new figure
sns.lineplot(x="Time (ms)", y="Sensor 1 (g)", data=myLab)
plt.title("Resonance Frequency vs Time")    
plt.xlabel("Time (ms)")
plt.ylabel("Acceleration (g)")
mplcursors.cursor(hover=False)  # Enable interactive cursor for data point selection
# Click anywhere on your data to add a cursor
# Right-click to remove a cursor
plt.show()
```
