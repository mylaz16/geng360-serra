## Displacement

$$\delta_{max} = \frac{FL^3}{3EI}$$

$$I = \frac{bh^3}{12}$$

```python
import numpy as np # You only need to import this in the first cell

F = 0.2205  # 0.5  # lb, Applied Force: 0.0110, 0.0220, 0.0441, 0.0661, 0.0882, 0.1102
L = 12.5  # 12.25  # in, Beam Length
E = 1e7  # psi, Young's Modulus for 6061 Aluminum
rho = 0.09754  # lb/in^3, density
b = 0.5  # in, beam width
h = 0.3  # in, beam height
ymax = h/2  # in, distance from neutral axis to outer fiber

I = b * h**3 / 12  # in^4, moment of inertia ; recall that in python ** is how you raise something to a power (not ^), rotating about the x-axis, thorugh the y-axis

delta = F * L**3 / (3 * E * I)

# f-string: {delta:.3g} shows delta with 3 significant figures (general format)
# other options: {delta:.3f} fixed 3 decimals, {delta:.2e} scientific, {delta:.0f} nearest int

print(f'Displacement is {delta:.3g} in') 
```

## Stress

$$\sigma = \frac{y_{max} M}{I} = \frac{y_{max} F  L}{I}$$

```python
F = 0.1102  # 0.5  # lb, Applied Force: 0.0110, 0.0220, 0.0441, 0.0661, 0.0882, 0.1102
L = 12.5  # 12.25  # in, Beam Length
h = 0.3  # in, beam height
ymax = h/2  # in, distance from neutral axis to outer fiber

I = b * h**3 / 12  # in^4, moment of inertia ; recall that in python ** is how you raise something to a power (not ^)

sigma = (ymax * F * L)/ I

# f-string: {delta:.3g} shows delta with 3 significant figures (general format)
# other options: {delta:.3f} fixed 3 decimals, {delta:.2e} scientific, {delta:.0f} nearest int
E = 1e7  # psi, Young's Modulus for 6061 Aluminum
rho = 0.09754  # lb/in^3, density
b = 0.5  # in, beam width

print(f'Stress is {sigma:.3g} psi') 
```

## Natural Frequency

$$f_n = \frac{K_n}{2\pi}\sqrt{\frac{EI}{\rho A L^4}}$$

```python
# Reminder to do these calculations in SI units,
# or else you have to account for a lbm to lbf
# unit conversion. No one wants to do that :) 
K = 3.52 # constant that depends on vibration mode
E = 1e7  # psi, Young's Modulus for 6061 Aluminum
rho = 0.09754  # lb/in^3, density
L = 12.5  # 12.25  # in, Beam Length
b = 0.5  # in, beam width
h = 0.3  # in, beam height

A = b * h  # in^2, cross-sectional area
I = b * h**3 / 12  # in^4, moment of inertia

f_n = K/(2*np.pi) * np.sqrt(E * I / (rho * A * L**4))  # in Hz, natural frequency of vibration for a cantilever beam 

print(f'Natural frequency is {f_n:.3g} Hz') 
```
