```python
fivedata = pd.read_csv("fivegramsdata.csv") # Loading the dataset  
fivedata.head() #Display first few rows  

filtered = fivedata[(fivedata['Times (us)'] >= 6e-6) & (fivedata['Times (us)'] <= 7.50e-6)] # Filter to cut out early data points  

V_ref = 5.06  # Reference voltage
filtered['Vout'] = (filtered['strain (raw)']/128) * (V_ref/16777215) # Convert raw strain data to voltage output using the provided conversion factor

Vr = (filtered['Vout']/V_ref)-(filtered['Vout'].mean()/V_ref)  # Convert Vout to Strain
GF = 2.11  # Gauge factor for the strain gauge
filtered['Strain'] = -10e6 * (-4 * Vr / (GF * (1+2 * Vr)))  # Calculate strain from voltage output and gauge factor

print(filtered['Strain'])
```
