```python
import serial
from pynmeagps import NMEAReader
import time

# Serial port configuration
PORT = 'COM3'  # Change to your GPS module's serial port (e.g., 'COM3' on Windows, '/dev/ttyUSB0' on Linux)
BAUD_RATE = 9600

# Initialize serial connection
try:
    ser = serial.Serial(PORT, BAUD_RATE, timeout=1)
    print(f"Connected to GPS module on {PORT}")
except serial.SerialException as e:
    print(f"Error connecting to serial port: {e}")
    ser = None

# Read and parse GPS data
if ser:
    try:
        nmea_reader = NMEAReader(ser)
        
        print("Reading GPS data... Press Ctrl+C to stop.")
        while True:
            try:
                msg, parsed = nmea_reader.read()
                
                # Check if message contains location data (GGA or RMC sentences)
                if parsed.msg_type in ['GGA', 'RMC']:
                    if hasattr(parsed, 'lat') and hasattr(parsed, 'lon'):
                        if parsed.lat and parsed.lon:
                            print(f"Latitude= {parsed.lat:.6f} Longitude= {parsed.lon:.6f}")
            except StopIteration:
                pass
            except Exception as e:
                print(f"Error reading data: {e}")
                
    except KeyboardInterrupt:
        print("\nStopped reading GPS data")
    finally:
        ser.close()
        print("Serial connection closed")
```
