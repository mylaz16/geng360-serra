```python
# import the dataset, view the first few rows
myData = pd.read_csv("PopUpAttempt1.csv") # Make sure to update this file name to match your data! 
myData.head() #This will display the first few rows of the DataFrame
```

I edited the code from the tutorial to show the y acceleration of the data set since we decided that the y-axis is the best to analyze and answer our research question. 

```python
# Creating sub plots for x and y accelerations
fig, ax = plt.subplots(1, 1, figsize=(7, 6))
sns.lineplot(x="Time (s)", y="Acceleration y (m/s^2)", data=myData,
              color = "red", linestyle="-", label="Acceleration y", ax=ax)


ax.set_title("Y Acceleration Plot") # Set title for the first plot
plt.xlabel("Time (s)")
ax.set_ylabel("Acceleration (m/s^2)") # Set y-label for the first plot

plt.xlim(1, 30)  # Set x-axis limits
plt.ylim(-40,60) # Set y-axis limits
plt.tight_layout()  # Adjust layout to prevent overlap
plt.show()
```
